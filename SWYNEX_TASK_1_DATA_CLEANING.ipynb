I AM MAKING COPY OF ORIGINAL DATA SO CLEANING PROCESS CAN BE PERFORMED ON IT

```python
c = df.copy()
c
```

```python
print(c.shape)
```

```python
c.isnull().sum()
```

```python
c.duplicated().sum()
```

Removing Duplicate Rows

```python
c = c.drop_duplicates().reset_index(drop=True)
c
```

```python
c['Name'].isnull().sum()
```

Changing Name Column to Standardize Format and Filling Null Values

```python
c["Name"] = c["Name"].str.strip().str.title()
c["Name"] = c["Name"].fillna("Unknown")
c['Name']
```

AGE

```python
c['Age'].isnull().sum()
```

```python
c['Age'].value_counts()
```

```python
c['Age'].replace("yrs","")
c['Age'] = pd.to_numeric(c['Age'], errors='coerce')
c.loc[c['Age'] < 0, 'Age'] = np.nan
c.loc[c['Age'] > 100, 'Age'] = np.nan
```

```python
c['Age']
```

```python
mean = c['Age'].mean()
round(mean)
```

```python
c['Age'].fillna(round(mean), inplace=True)
c['Age']
```

```python
c['Age'].value_counts()
```

Gender

```python
c['Gender']
```

```python
c['Gender'].value_counts()
```

```python
c.Gender= c.Gender.replace(['male','m','M','MALE'],"Male")
c.Gender= c.Gender.replace(['female','f','FEMALE','F'],"Female")
c.Gender
```

```python
c['Gender'].isnull().sum()
```

```python
c['Gender'] = c['Gender'].fillna("Not Identified")
c['Gender']
```

```python
c['Gender'].isnull().sum() # Now Null is 0
```

```python
c["Gender"].value_counts()
```

City

```python
c['City']
```

```python
c['City'].value_counts()
```

replace() = change values \
map() = convert values according to a mapping/dictionary.

```python
city_map = {
    "bombay": "Mumbai",
    "mumbai": "Mumbai",
    "calcutta": "Kolkata",
    "kolkata": "Kolkata",
    "madras": "Chennai",
    "chennai": "Chennai",
    "banaras": "Varanasi",
    "varanasi": "Varanasi",
    "bangalore": "Bengaluru",
    "bengaluru": "Bengaluru",
    "delhi": "Delhi",
    "new delhi": "Delhi",
    "hyderabad": "Hyderabad",
    "pune": "Pune",
}
c["City"] = c["City"].str.strip().str.lower().map(city_map)
c['City'] = c['City'].fillna("Unknown")
c['City']
```

```python
c['City'].value_counts()
```

JoinDate

```python
c['JoinDate']
```

```python
c['JoinDate'].isnull().sum()
```

```python
c['JoinDate'].value_counts()
```

```python
c.JoinDate = pd.to_datetime(c['JoinDate'], format = "mixed")
c.JoinDate
```

```python
c["JoinDate"].median()
```

```python
c["JoinDate"].fillna(c["JoinDate"].median(), inplace=True)
c["JoinDate"]
```

PurchaseAmount

```python
c['PurchaseAmount']
```

```python
c['PurchaseAmount'].isnull().sum()
```

```python
c['PurchaseAmount']= c['PurchaseAmount'].str.replace("₹","")
c['PurchaseAmount'] = c['PurchaseAmount'].str.replace(",","")
c['PurchaseAmount'] = c['PurchaseAmount'].astype(float)
c['PurchaseAmount']
```

```python
c["PurchaseAmount"] = pd.to_numeric(c["PurchaseAmount"], errors="coerce")
c["PurchaseAmount"]
```

```python
c["PurchaseAmount"] = (c["PurchaseAmount"].fillna(c["PurchaseAmount"].median()).round(2))
c['PurchaseAmount']
```

ProductCategory

```python
c['ProductCategory']
```

```python
c['ProductCategory'].isnull().sum()
```

```python
c['ProductCategory'].value_counts()
```

```python
category_map = {
    "book": "Books",
    "books": "Books",
    "clothing": "Clothing",
    "apparel": "Clothing",
    "electronics": "Electronics",
    "electronic": "Electronics",
    "groceries": "Groceries",
    "grocery": "Groceries",
    "furniture": "Furniture",
}
c["ProductCategory"] = (
    c["ProductCategory"].astype(str).str.strip().str.lower().map(category_map)
)
c["ProductCategory"] = c["ProductCategory"].fillna("Other")
c['ProductCategory']
```

Rating

```python
c['Rating']
```

```python
c['Rating'].median()
```

```python
# Invalid ratings → NaN
c.loc[c["Rating"] > 5, "Rating"] = np.nan

# Fill missing ratings + round
c["Rating"] = c["Rating"].fillna(c["Rating"].median()).round(1)
c['Rating']
```
